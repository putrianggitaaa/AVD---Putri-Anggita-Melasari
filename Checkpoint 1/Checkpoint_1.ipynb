# Business Understanding

Sebuah Universitas memiliki Dataset bernama Student Depression. Dari dataset ini, Universitas tersebut ingin melakukan analisis terhadap faktor-faktor penyebab depresi pada pelajar dan juga solusi untuk mengatasi permasalahan tersebut. Oleh karena itu, berikut saya akan menjabarkan 4 tahapan Business Understanding pada studikasus ini.

## Business Objective ##


Tujuan bisnis dari dataset ini adalah untuk mengetahui faktor-faktor penyebab pelajar mengalami depresi dalam menempuh pendidikan. Hal ini membantu Universitas atau pihak terkait untuk mengambil langkah selanjutnya seperti pendampingan mental bagi pelajar di tiap-tiap instansi pendidikan dengan cara bimbingan konseling yang optimal, monitoring kinerja tenaga pengajar secara berkala, bantuan finansial bagi pelajar dengan tingkat ekonomi menengah kebawah, edukasi kepada orang tua pelajar mengenai bahaya depresi, perbaikan kurikulum, dan berbagai langkah pencegahan lainnya demi kesejahteraan sumber daya manusia dan juga peningkatan kualitas pendidikan dunia secara berkelanjutan.




## Assess Situation ##

Seperti yang kita tahu, depresi dini pada pelajar dapat mempengaruhi keberhasilan akademik dan kesejahteraan hidup mereka sebagai seseorang yang masih harus mengemban tugas dan kewajiban panjang sebagai seorang pelajar. Tekanan yang berlebih pada pelajar sangat berakibat fatal. Oleh karena itu, data ini perlu dianalisis untuk memahami secara jelas faktor-faktor penyebabnya.

Student Depression Dataset berisi 27.901 data pelajar yang dimana masing-masing data tersebut memiliki 18 kolom variabel berbeda mengenai faktor depresi pelajar, yaitu: tekanan akademik, IPK, kepuasan dalam belajar, durasi tidur, kebiasaan diet, jenjang pendidikan yang diambil, pernah atau tidaknya memiliki niat untuk bunuh diri, jam belajar, tekanan finansial, dan riwayat kesehatan mental genetik.

Analisis dataset ini tidak serta merta berhasil dijalankan. Ada beberapa hal yang perlu menjadi perhatian dalam proses analisis, seperti data-data pelajar yang ada berasal dari usia dan juga jenjang pendidikan yang berbeda. Hal ini membuat proses perbandingan menjadi cukup sulit karena tiap-tiap jenjang pendidikan memiliki tekanan yang berbeda. Selain itu, beberapa kolom numerik seperti tekanan akademik dan IPK memiliki skala yang berbeda sehingga memicu perbedaan interpretasi hasil analisis data nantinya.

## Analytic Goals ##

Tujuan analitik dataset ini, yaitu sebagai berikut:

1.   Menganalisis korelasi antara pola hidup pelajar seperti pola makan, jam tidur, dan waktu belajar yang berlebih dengan kondisi depresi.
2.   Mengukur seberapa besar pengaruh hubungan antara tingkat tekanan akademik dengan kemungkinan mahasiswa mengalami depresi.
3. Menganalisis perbedaan tingkat depresi berdasarkan jenjang dan bidang pendidikan.
4. mengidentifikasi apakah tekanan ekonomi dan riwayat penyakit mental dalam keluarga meningkatkan risiko depresi.
5. mengidentifikasi hubungan antara performa akademik yaitu IPK dengan risiko depresi.
6. menganalisis hubungan antara kepuasan belajar dengan kondisi kesehatan mental.
